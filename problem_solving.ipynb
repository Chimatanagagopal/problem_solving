```python
numbers = [10, 5, 20, 8, 20, 15, 3]
first_max = float("-inf")
second_max = float("-inf")
for i in numbers:
  if i>first_max:
    second_max=first_max
    first_max=i
  elif i>second_max and i!= first_max:
    second_max=i
print(second_max)
```

```python
# Problem 2 — Find the first non-repeating element
```

```python

numbers = [9,4, 5, 1, 2, 1, 4, 5, 2, 7]
for i in range(len(numbers)):
  for j in range(len(numbers)):
    if i!=j and numbers[i]==numbers[j]:
      break
  else:
    print(numbers[i])
    break
```

```python
numbers = [1, 2, 3, 5, 6, 7, 8]
for num in range(len(numbers)):
  if numbers[num+1]!=numbers[num]+1:
    print(numbers[num]+1)
    break
```
