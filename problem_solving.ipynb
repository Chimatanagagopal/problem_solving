```python
numbers = [10, 5, 20, 8, 20, 15, 3]
first_max = float("-inf")
second_max = float("-inf")
for i in numbers:
  if i>first_max:
    second_max=first_max
    first_max=i
  elif i>second_max and i!= first_max:
    second_max=i
print(second_max)
```

```python
# Problem 2 — Find the first non-repeating element
```

```python

numbers = [9,4, 5, 1, 2, 1, 4, 5, 2, 7]
for i in range(len(numbers)):
  for j in range(len(numbers)):
    if i!=j and numbers[i]==numbers[j]:
      break
  else:
    print(numbers[i])
    break
```

```python
# Find the missing number
```

```python
numbers = [1, 2, 3, 5, 6, 7, 8]
for num in range(len(numbers)- 1):
  if numbers[num+1]!=numbers[num]+1:
    print(numbers[num]+1)
    break
```

```python
# Problem 4 — Find duplicate elements
```

```python
numbers = [4, 2, 7, 4, 2, 8, 9, 7, 7]
duplicates=[]
for i in range(len(numbers)):
  for j in range(len(numbers)):
    if i != j and numbers[i]==numbers[j] and numbers[i] not in duplicates:
      duplicates+=[numbers[i]]
print(duplicates)
```

```python
#Problem 5 — Find the first pair whose sum equals the target
```

```python
numbers = [1, 5, 3, 6]
target = 9
found = False
for i in range(len(numbers)):
  for j in range(i+1,len(numbers)):
    if numbers[i]+numbers[j] ==target:
      print(numbers[i],numbers[j])
      found = True
      break
  if found:
    break
```

```python
#Problem 6 — Largest difference
```

```python
numbers = [10, 3, 8, 20, 5, 15]
largest_number= float("-inf")
smallest_number= float("inf")
for i in numbers:
  if largest_number <i:
    largest_number=i
  if smallest_number >i:
    smallest_number=i
print(largest_number-smallest_number)
```
